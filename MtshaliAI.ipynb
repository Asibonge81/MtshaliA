```python

def detect_outliers(df, columns):
    for col in columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)

        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        outliers = df[
            (df[col] < lower) |
            (df[col] > upper)
        ]

        print(f"{col}: {len(outliers)} potential outliers")

detect_outliers(df, numeric_cols1)
detect_outliers(df2, numeric_cols2)
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# Exact crime category names used in SAPS
categories = [
    "Rape", "Sexual assault", "Kidnapping",
    "Murder", "Attempted murder", "Common assault",
    "Assault with the intent to inflict grievous bodily harm",
    "Robbery with aggravating circumstances"
]

def prepare(data, quarter):
    data = data.copy()
    data.columns = data.columns.map(lambda x: str(x).strip())

    data = data[
        data["District"].str.contains("Ekurhuleni", case=False, na=False)
        & data["Crime_Category"].isin(categories)
    ]

    year_cols = [
        c for c in data.columns
        if c.startswith(("January", "April"))
        and "to" in c
    ]

    data = data.melt(
        id_vars=["Station", "Crime_Category"],
        value_vars=year_cols,
        var_name="Period",
        value_name="Cases"
    )

    data["Year"] = data["Period"].str.extract(r"(20\d{2})").astype(int)
    data["Quarter"] = quarter
    data["Cases"] = pd.to_numeric(data["Cases"], errors="coerce")

    return data.dropna(subset=["Cases"])

jan = prepare(df, "Jan-Mar")
apr = prepare(df2, "Apr-Jun")

print("January-March records:", len(jan))
print("April-June records:", len(apr))
```

# **Modelling**

I chose the random Forest Classifier to  predict whether certain areas near police-station in Ekurhuleni will be classified as a crime hotspot, so that the results can support the prioritisation of prevention resources and safety planning.

```python

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import balanced_accuracy_score
from sklearn.model_selection import train_test_split

import numpy as np

# Combine the two quarters for modelling
data = pd.concat([jan, apr], ignore_index=True)

# Total selected reported crimes per station and quarter
data = data.groupby(
    ["Station", "Year", "Quarter"],
    as_index=False
)["Cases"].sum()

# Define hotspots: top 25% within each year and quarter
data = data.sort_values(["Year", "Quarter", "Station"])

data["Hotspot"] = (
    data.groupby(["Year", "Quarter"])["Cases"]
    .transform(lambda x: x.rank(method="first", ascending=False)
               <= np.ceil(len(x) * 0.25))
    .astype(int)
)

print(data["Hotspot"].value_counts())
display(data.head())
```

```python

data = data.sort_values(["Station", "Quarter", "Year"])

groups = data.groupby(["Station", "Quarter"])

data["Next_Hotspot"] = groups["Hotspot"].shift(-1)
data["Next_Year"] = groups["Year"].shift(-1)

# Keep only consecutive-year predictions
ml_df = data[
    data["Next_Year"] == data["Year"] + 1
].copy()

X = ml_df[["Station", "Quarter", "Year", "Cases", "Hotspot"]]
y = ml_df["Next_Hotspot"].astype(int)

X = pd.get_dummies(X, columns=["Station", "Quarter"])

# Train on earlier years, test predictions for 2026
train = ml_df["Next_Year"] <= 2025
test = ml_df["Next_Year"] == 2026

X_train, X_test = X.loc[train], X.loc[test]
y_train, y_test = y.loc[train], y.loc[test]

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Target distribution:")
print(y_train.value_counts())
```

```python

model = RandomForestClassifier(
    n_estimators=200,
    max_depth=8,
    min_samples_leaf=3,
    class_weight="balanced",
    random_state=42
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Model trained successfully!")
```

```python

print(classification_report(
    y_test, y_pred,
    target_names=["Not Hotspot", "Hotspot"],
    zero_division=0
))

print("Balanced Accuracy:",
      round(balanced_accuracy_score(y_test, y_pred), 3))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python

results = ml_df.loc[test, [
    "Station", "Quarter", "Next_Year"
]].copy()

results["Actual_Hotspot"] = y_test.values
results["Predicted_Hotspot"] = y_pred
results["Hotspot_Probability"] = (
    model.predict_proba(X_test)[:, 1]
)

results = results.sort_values(
    "Hotspot_Probability", ascending=False
)

display(results.head(15))
```

# **Pattern Discovery and Spatial/Temporal Analysis**

 For the 3 hotspots I chose the ones with the highest combined crime counts.

 The purpose of this analysis was to investigate crime patternsi Ekurhuleni and understand the sismilarities betweeen 3 selected hotspot areas.

 It focused on mainly 3 patterns :
 1. Changes over time
 2. Recurring patterns
 3. Spatial Concentration

 The 3 areas were also compared based on their crime categorie and historical trends.

 **Association versus causation**
 Although two areas may expirience simila crime patterns this does not mean they have the same causes. For example, higher reported crime counts may be associated with population size or reporting practices, but the available data cannot prove that these factors caused the increase.

```python

# Combine the quarterly data for analysis
both = pd.concat([jan, apr], ignore_index=True)

# Select the three highest-count areas in 2026
top3 = (
    both[both["Year"] == 2026]
    .groupby("Station")["Cases"]
    .sum()
    .nlargest(3)
)

stations = top3.index.tolist()

print("Three selected hotspot areas:")
print(top3)
```

```python

trend = (
    both[both["Station"].isin(stations)]
    .groupby(["Station", "Year"], as_index=False)["Cases"]
    .sum()
)

plt.figure(figsize=(10, 5))

sns.lineplot(
    data=trend,
    x="Year",
    y="Cases",
    hue="Station",
    marker="o",
    linewidth=2.5
)

plt.title("Crime Trends in Three Ekurhuleni Hotspot Areas")
plt.xlabel("Year")
plt.ylabel("Reported Cases (January-June)")
plt.xticks(range(2022, 2027))
plt.tight_layout()
plt.show()
```

```python

quarterly = both[both["Station"].isin(stations)].pivot_table(
    index=["Station", "Year"],
    columns="Quarter",
    values="Cases",
    aggfunc="sum"
).reset_index()

quarterly["Difference"] = (
    quarterly["Apr-Jun"] - quarterly["Jan-Mar"]
)

plt.figure(figsize=(9, 6))

sns.barplot(
    data=quarterly,
    x="Year",
    y="Difference",
    hue="Station"
)

plt.axhline(0, color="black", linewidth=1)
plt.title("Quarterly Crime Differences by Hotspot Area")
plt.ylabel("Apr-Jun Cases Minus Jan-Mar Cases")
plt.xlabel("Year")
plt.tight_layout()
plt.show()

## Similar-shaped lines indicate that areas experienced similar changes over time.
#For example, if two stations show increases during the same years, they share a temporal pattern. However, that does not establish that the increases had the same cause.
```

```python

crime_mix = both[
    (both["Station"].isin(stations)) &
    (both["Year"] == 2026)
].pivot_table(
    index="Station",
    columns="Crime_Category",
    values="Cases",
    aggfunc="sum",
    fill_value=0
)

crime_pct = crime_mix.div(crime_mix.sum(axis=1), axis=0) * 100

crime_pct.plot(
    kind="bar",
    stacked=True,
    figsize=(11, 6),
    colormap="tab20"
)

plt.title("Crime Category Profiles of Three Hotspot Areas")
plt.xlabel("Police Station")
plt.ylabel("Percentage of Selected Reported Cases")
plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python

from sklearn.linear_model import LinearRegression

records = []

for station in stations:
    d = both[both["Station"] == station].groupby(
        ["Year", "Quarter"], as_index=False
    )["Cases"].sum()

    X_pattern = pd.DataFrame({
        "Year": d["Year"] - d["Year"].min(),
        "AprJun": (d["Quarter"] == "Apr-Jun").astype(int)
    })

    model_pattern = LinearRegression()
    model_pattern.fit(X_pattern, d["Cases"])

    records.append({
        "Station": station,
        "Annual_Change": round(model_pattern.coef_[0], 2),
        "Quarter_Difference": round(model_pattern.coef_[1], 2),
        "R2": round(model_pattern.score(X_pattern, d["Cases"]), 3)
    })

pattern_results = pd.DataFrame(records)
display(pattern_results)
```

Hotspot Pattern Analysis and Comparison
The purpose of this analysis was to investigate crime patterns in Ekurhuleni and understand the similarities and differences between three selected hotspot areas.

Using historical crime statistics from 2022 to 2026, three police-station areas with high reported crime counts were selected for deeper comparison.

The analysis focused on three main patterns:

1. Changes over time: Crime trends were examined to determine whether reported cases increased, decreased or remained stable over the years.

2. Recurring patterns: Crime statistics from January–March and April–June were compared to identify whether certain periods consistently recorded higher crime levels.

3. Spatial concentration: Reported crime counts were compared across police-station areas to identify where selected offences were most concentrated.

The three areas were also compared based on their crime categories and historical trends. Areas showing similar increases, recurring quarterly differences or similar crime distributions were considered to have similar patterns.

Association versus causation: Although two areas may experience similar crime patterns, this does not mean they have the same causes. For example, higher reported crime counts may be associated with population size or reporting practices, but the available data cannot prove that these factors caused the increase.

Conclusion: This analysis helps decision-makers understand crime patterns beyond simply identifying hotspots. The findings can support the prioritisation of safety initiatives, prevention resources and further investigations in Ekurhuleni. However, the statistics represent reported crimes generally and should not be interpreted as women-specific victimisation rates.
